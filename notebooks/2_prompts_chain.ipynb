#### Prompt Templates - Dynamic Prompts

```python
from langchain_openrouter import ChatOpenRouter
from langchain_core.prompts import ChatPromptTemplate
```

```python
llm = ChatOpenRouter(model="cohere/north-mini-code:free")
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role":"system","content":"You are a {role}."},
    {"role":"user","content":"Give me the best {niche} startup idea to start now in India in 2026?"}
])
```

```python
prompts_final = prompts.format_messages(niche = "Sex", role = "Y Combinator Indian leading top expert, who knows the market thoroughly.")
#prompts_final
res = llm.invoke(prompts_final)
print(res.content)
```

#### Chaining

```python
final = prompts.invoke({"niche":"Batman", "role":"Y Combinator Indian leading top expert, who knows the market thoroughly."})
res = llm.invoke(final)
print(res.content)
```

##### Best and the shortcut method

```python
Chains = prompts | llm
```

```python
res = Chains.invoke({"niche":"Donald Trump", "role":"Jeffrey Epstein"})
print(res)
```

### Bestest Method nowwww

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
out = StrOutputParser()
```

```python
prompts1 = ChatPromptTemplate.from_messages([
    {"role":"system", "content":"You are a translator and translate input in {language}"},
    {"role":"user", "content":"{query}"},
])
```

```python
### Defining my own chain
def transform_case(result:str):
    return result.upper()
```

```python
Chains = prompts1 | llm | out | transform_case
```

```python
res = Chains.invoke({"language":"Hinglish","query":"I would love to be a billionaire"})
print(res)
```
