```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openrouter import ChatOpenRouter
llm = ChatOpenRouter(model="nvidia/nemotron-3-super-120b-a12b:free")
```

```python
prompts = [
    {"role":"user", "content":"Hello, my name is Soham Ray."},
    {"role":"ai", "content":"Hello, Soham Ray! It's nice to meet you. How can I help you today?"},
    {"role":"user", "content":"What is my name?"}
]

res = llm.invoke(prompts)
```

```python
print(res.content)
```

### Manage History

```python
history = []
while True:
    query = input("User: ")
    if query.lower() in ["exit", "quit", "bye", "end", "stop"]:
        print("Goodbye!👋")
        break

    history.append({"role":"user", "content":query})
    print("User: ", query)

    res = llm.invoke(history)
    history.append({"role":"ai", "content":res.content})
    print("AI: ", res.content, "\n")
```
